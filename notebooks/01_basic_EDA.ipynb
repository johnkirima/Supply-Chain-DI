**NOW WE LOAD THE DATA**



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set visual style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = [12, 6]

# Define paths
DATA_DIR = "../data"

# Load the data
products = pd.read_csv(os.path.join(DATA_DIR, "products.csv"))
warehouses = pd.read_csv(os.path.join(DATA_DIR, "warehouses.csv"))
shipping_rates = pd.read_csv(os.path.join(DATA_DIR, "shipping_rates.csv"))
demand_history = pd.read_csv(os.path.join(DATA_DIR, "demand_history.csv"))

# Convert date to datetime
demand_history['date'] = pd.to_datetime(demand_history['date'])

print("All files loaded successfully!")
print(f"Demand History Shape: {demand_history.shape}")
```
